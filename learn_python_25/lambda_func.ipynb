<h1>Lambda Functions</h1>

```python
def square(x):
    return x**2
```

```python
square(10)
```

```python
# concise function using lambda
square = lambda x : x**2
```

```python
type(square)
```

```python
square(10)
```

```python
def random(x):
    return(x + 10)
```

```python
random(20)
```

```python
lambda_random = lambda x : x+10
```

```python
lambda_random(20)
```

```python
# anonymous function
(lambda x:x+10)(20)
```

```python
(lambda x,y:x+10+y)(20,15)
```
